```python
import yfinance as yf

raw_data = yf.download(["SPY", "GLD"], start="2016-01-01", end="2026-11-25", auto_adjust=True)
close_prices = raw_data["Close"].dropna()
close_prices.columns = ["GLD_Close", "SPY_Close"]
close_prices.to_csv("market_data.csv")

print(f"Kész! {len(close_prices)} kereskedési nap mentve.")
close_prices.head()
```
